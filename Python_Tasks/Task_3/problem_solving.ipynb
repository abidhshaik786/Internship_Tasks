### Q1. Find the Frequency of Each Word
1. Take a sentence as input.
2. Convert the sentence to lowercase and split it into words.
3. Create an empty dictionary.
4. Count how many times each word appears.
5. Return the dictionary.


```python
def word_frequency(sentence):
    frequency = {}

    words = sentence.lower().split()

    for word in words:
        if word in frequency:
            frequency[word] += 1
        else:
            frequency[word] = 1

    return frequency


# Test case 1
print(word_frequency("Python is easy and Python is powerful"))

# Test case 2
print(word_frequency("hello hello world"))
```

### Q2. Find the Frequency of Each Character

**Pseudocode:**
1. Take a word as input.
2. Create an empty dictionary.
3. Count how many times each character appears.
4. Return the character frequencies.

```python
def character_frequency(word):
    frequency = {}
    for char in word:
        if char in frequency:
            frequency[char] += 1
        else:
            frequency[char] = 1

    return frequency

# Test case 1
print(character_frequency("banana"))

# Test case 2
print(character_frequency("apple"))


```

### Q3. Find the First Non-Repeating Character

**Pseudocode:**
1. Take a string as input.
2. Check each character in the string.
3. If a character appears only once, return it.
4. If no such character exists, return None.

```python
def first_non_repeating(word):
    for char in word:
        if word.count(char) == 1:
            return char
    return None

# Test case 1
print(first_non_repeating("swiss"))

# Test case 2
print(first_non_repeating("aabbc"))
```

### Q4. Check Whether Two Strings Are Anagrams

**Pseudocode:**
1. Take two strings as input.
2. Sort the characters of both strings.
3. Compare the sorted strings.
4. Return True if they match; otherwise, return False.

```python
def are_anagrams(s1, s2):
    if sorted(s1) == sorted(s2):
        return True
    else:
        return False

# Test case 1
print(are_anagrams("listen", "silent"))

# Test case 2
print(are_anagrams("hello", "world"))
```

### Q5. Find the First Repeating Character

**Pseudocode:**
1. Take a string as input.
2. Create an empty list to store seen characters.
3. Check each character and return the first one already seen.
4. If no character repeats, return None.

```python
def first_repeating(string):
    seen = []
    for char in string:
        if char in seen:
            return char
        else:
            seen.append(char)

    return None

# Test case 1
print(first_repeating("programming"))

# Test case 2
print(first_repeating("abcdefa"))
```

### Q6. Move All Zeros to the End

**Pseudocode:**
1. Take a list of numbers.
2. Add all non-zero numbers to a new list.
3. Add all zeros to the end.
4. Return the new list.

```python
def move_zeros(nums):
    result = []
    for num in nums:
        if num != 0:
            result.append(num)
    for num in nums:
        if num == 0:
            result.append(num)

    return result

# Test case 1
print(move_zeros([0, 3, 0, 1, 2, 0, 4]))

# Test case 2
print(move_zeros([1, 0, 2]))
```

### Q7. Find the Second-Highest Number

**Pseudocode:**
1. Take a list of numbers.
2. Remove duplicate numbers.
3. Sort the numbers in descending order.
4. Return the second-highest number, or None if it does not exist.

```python
def second_highest(nums):
    unique_nums = list(set(nums))
    unique_nums.sort(reverse=True)
    if len(unique_nums) < 2:
        return None

    return unique_nums[1]

# Test case 1
print(second_highest([10, 20, 40, 30, 40, 20]))

# Test case 2
print(second_highest([5, 5, 3, 1]))
```

### Q8. Remove Duplicates from a List

**Pseudocode:**
1. Take a list of numbers.
2. Create an empty list.
3. Add a number only if it is not already in the new list.
4. Return the list without duplicates.

```python
def remove_duplicates(nums):
    result = []
    for num in nums:
        if num not in result:
            result.append(num)

    return result

# Test case 1
print(remove_duplicates([4, 2, 4, 5, 2, 7, 5]))

# Test case 2
print(remove_duplicates([1, 1, 2, 3, 2]))
```

### Q9. Find the Missing Number

**Pseudocode:**
1. Take a list containing numbers from 1 to N, with one number missing.
2. Calculate the expected sum from 1 to N.
3. Calculate the sum of the given numbers.
4. Subtract the actual sum from the expected sum to find the missing number.

```python
def missing_number(nums, n):
    expected_sum = n * (n + 1) // 2
    actual_sum = sum(nums)

    return expected_sum - actual_sum

# Test case 1
print(missing_number([1, 2, 4, 5, 6], 6))

# Test case 2
print(missing_number([1, 3, 4], 4))
```

### Q10. Find Two Numbers Whose Sum Equals the Target

**Pseudocode:**
1. Take a list of numbers and a target value.
2. Check each number and calculate the required number.
3. Check whether the required number was seen earlier.
4. Return the pair if found; otherwise, return an empty list.

```python
def two_sum(nums, target):
    seen = []
    for num in nums:
        needed = target - num
        if needed in seen:
            return [needed, num]
        seen.append(num)

    return []

# Test case 1
print(two_sum([2, 7, 11, 15], 9))

# Test case 2
print(two_sum([3, 2, 4], 6))
```

### Q11. Find the Number Closest to Zero

**Pseudocode:**
1. Take a list of positive and negative numbers.
2. Start with the first number as the closest number.
3. Compare the absolute values of the numbers.
4. Return the number closest to zero; if there is a tie, return the positive number.

```python
def closest_to_zero(nums):
    closest = nums[0]
    for num in nums:
        if abs(num) < abs(closest):
            closest = num
        elif abs(num) == abs(closest) and num > closest:
            closest = num

    return closest

# Test case 1
print(closest_to_zero([-8, -3, 2, 5, -2]))

# Test case 2
print(closest_to_zero([-5, 3, -1]))
```

### Q12. Find Common Elements in Two Lists

**Pseudocode:**
1. Take two lists of numbers.
2. Check each number in the first list.
3. Add it if it exists in the second list and is not already added.
4. Return the common elements.

```python
def common_elements(list1, list2):
    result = []
    for num in list1:
        if num in list2 and num not in result:
            result.append(num)

    return result

# Test case 1
print(common_elements([1, 2, 3, 4, 5], [3, 4, 5, 6, 7]))

# Test case 2
print(common_elements([4, 2, 4, 1], [4, 1, 4]))
```

### Q13. Rotate a List to the Right

**Pseudocode:**
1. Take a list and the number of rotations.
2. Handle an empty list.
3. Use the list length to handle rotations larger than the list.
4. Move the last K elements to the beginning and return the result.

```python
def rotate_right(nums, k):
    if len(nums) == 0:
        return []
    k = k % len(nums)
    if k == 0:
        return nums[:]
    result = nums[-k:] + nums[:-k]

    return result

# Test case 1
print(rotate_right([1, 2, 3, 4, 5], 2))

# Test case 2
print(rotate_right([10, 20, 30], 1))
```

### Q14. Find the Maximum Sum of Two Consecutive Numbers

**Pseudocode:**
1. Take a list containing at least two numbers.
2. Add the first two numbers and store the sum.
3. Add each pair of consecutive numbers and compare their sums.
4. Return the maximum sum.

```python
def max_consecutive_sum(nums):
    max_sum = nums[0] + nums[1]
    for i in range(len(nums) - 1):
        total = nums[i] + nums[i + 1]
        if total > max_sum:
            max_sum = total

    return max_sum

# Test case 1
print(max_consecutive_sum([4, 7, 2, 9, 5]))

# Test case 2
print(max_consecutive_sum([1, 2, 3, 4]))
```

### Q15. Find All Pairs Whose Sum Equals the Target

**Pseudocode:**
1. Take a list of numbers and a target value.
2. Check every possible pair of different positions.
3. If a pair adds up to the target, add it to the result without duplicates.
4. Sort the pairs and return the result.

```python
def all_pairs(nums, target):
    pairs = []
    for i in range(len(nums)):
        for j in range(i + 1, len(nums)):
            if nums[i] + nums[j] == target:
                pair = [nums[i], nums[j]]
                pair.sort()
                if pair not in pairs:
                    pairs.append(pair)

    pairs.sort()

    return pairs

# Test case 1
print(all_pairs([2, 4, 3, 5, 7, 8, 1], 9))

# Test case 2
print(all_pairs([1, 1, 2, 2, 3, 4], 5))
```
