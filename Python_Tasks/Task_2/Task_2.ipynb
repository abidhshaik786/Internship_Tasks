```python
#Q2 Ans,

# Step 1: Observe
# Input: [12,45,7,89,23]
# output: 89

# Step 2: Understand
# Find the largest number without using max().

# Step 3: Break
# Start with the first number and compare the remaining numbers.

# Step 4: Logic
# Update largest whenever a bigger number is found.

# Step 5: Code
numbers = [12,45,7,89,23]

largest = numbers[0]
for number in numbers:
    if number > largest:
        largest = number
print(largest)
    
# Step 6: Test
test_numbers = [4, 15, 2, 9]
test_largest = test_numbers[0]
for number in test_numbers:
    if number > test_largest:
        test_largest = number
print(test_largest)
```

```python
#Q4 Ans,

# Step 1: Observe
# Input: "Python makes coding useful"
# output: "Python"

# Step 2: Understand
# Find the longest word, keeping the first word in case of a tie.

# Step 3: Break
# Split the sentence, compare word lengths, and select the longest.

# Step 4: Logic
# Update the longest word only when a strictly longer word is found.

# Step 5: Code
sentence = "Python makes coding useful"
words = sentence.split()
longest_word = words[0]
for word in words:
    if len(word) > len(longest_word):
        longest_word = word
print(longest_word)

# Step 6: Test
test_sentence = "cat tiger rabbit"
test_words = test_sentence.split()
test_longest = test_words[0]

for word in test_words:
    if len(word) > len(test_longest):
        test_longest = word

print(test_longest)
```

```python
#Q7 Ans,

# Step 1: Observe
# Input: "Python 123!"
# output: 5

# Step 2: Understand
# Count consonants and ignore digits, spaces, and special characters.

# Step 3: Break
# Check whether each character is a letter and not a vowel.

# Step 4: Logic
# Count alphabetic characters that are not vowels.

# Step 5: Code
text = "Python 123!"
vowels = "aeiouAEIOU"
count = 0
for char in text:
    if char.isalpha() and char not in vowels:
        count +=1
print(count)

# step 6: Test
test_text = "Hello 45!"
test_count = 0
for character in test_text:
    if character.isalpha() and character not in vowels:
        test_count += 1
print(test_count)
```

```python
#Q8 Ans,

# Step 1: Observe
# Input: "banana"
# output: "a"

# Step 2: Understand
# Find the character occurring most often, ignoring spaces.

# Step 3: Break
# Count character frequencies and track the highest.

# Step 4: Logic
# Update the answer only when a higher frequency is found.

# Step 5: Code
text = "banana"
text = text.replace(" ", "")
most = text[0]
for ch in text:
    if text.count(ch) > text.count(most):
        most = ch
print(most)

# step 6: Test
text = "apple"
text = text.replace(" ", "")
most = text[0]
for ch in text:
    if text.count(ch) > text.count(most):
        most = ch
print(most)
```

```python
#Q10 Ans,

# Step 1: Observe
# Input: "programming"
# output: ['r', 'g', 'm']

# Step 2: Understand
# Find repeated characters, ignoring spaces.

# Step 3: Break
# Check frequencies and store each repeated character once.

# Step 4: Logic
# Add a character if its frequency is greater than one
# and it is not already in the result.

# Step 5: Code
text = "programming"
text = text.replace(" ", "")
repeated = []
for character in text:
    if text.count(character) > 1 and character not in repeated:
        repeated.append(character)
print(repeated)

# Step 6: Test
test_text = "hello world"
test_text = test_text.replace(" ", "")
test_repeated = []
for character in test_text:
    if test_text.count(character) > 1 and character not in test_repeated:
        test_repeated.append(character)
print(test_repeated)
```

```python
#Q15 Ans,

# Step 1: Observe
# Input: [1, 2, 3, 4, 5]
# output: {1: 'Odd', 2: 'Even', 3: 'Odd', 4: 'Even', 5: 'Odd'}

# Step 2: Understand
# Identify each number as even or odd.

# Step 3: Break
# Check the remainder when each number is divided by 2.

# Step 4: Logic
# Remainder 0 means even; otherwise, odd.

# Step 5: Code
numbers = [1, 2, 3, 4, 5]
result = {}
for number in numbers:
    if number % 2 == 0:
        result[number] = "Even"
    else:
        result[number] = "Odd"
print(result)

# Step 6: Test
test_numbers = [10, 13]
test_result = {}
for number in test_numbers:
    if number % 2 == 0:
        test_result[number] = "Even"
    else:
        test_result[number] = "Odd"
print(test_result)
```

```python
#Q16 Ans,

# Step 1: Observe
# Input: "Python is easy and Python is powerful"
# output: "python"

# Step 2: Understand
# Count words without distinguishing uppercase and lowercase.

# Step 3: Break
# Convert to lowercase, split into words, and count frequencies.

# Step 4: Logic
# Keep the first word with the highest frequency.

# Step 5: Code
text = "Python is easy and Python is powerful"
text = text.lower()
words = text.split()
most = words[0]
for word in words:
    if words.count(word) > words.count(most):
        most = word
print(most)
# Step 6: Test
text = "Cat dog cat bird dog cat"
text = text.lower()
words = text.split()
most = words[0]
for word in words:
    if words.count(word) > words.count(most):
        most = word
print(most)
```

```python
#Q17 Ans,

# Step 1: Observe
# Input: "data science uses data and science tools"
# output: ['data', 'science']

# Step 2: Understand
# Find repeated words regardless of capitalization.

# Step 3: Break
# Convert to lowercase, split, and check each word.

# Step 4: Logic
# Add repeated words once, in first-appearance order.

# Step 5: Code
sentence = "data science uses data and science tools"
words = sentence.lower().split()
repeated = []
for word in words:
    if words.count(word) > 1 and word not in repeated:
        repeated.append(word)
print(repeated)

# Step 6: Test
test_sentence = "Apple banana APPLE orange"
test_words = test_sentence.lower().split()
test_repeated = []
for word in test_words:
    if test_words.count(word) > 1 and word not in test_repeated:
        test_repeated.append(word)
print(test_repeated)
```

```python
#Q18 Ans,

# Step 1: Observe
# Input: "apple banana apple mango banana orange"
# output: "mango"

# Step 2: Understand
# Find the first word occurring exactly once, ignoring case.

# Step 3: Break
# Convert to lowercase, split, and check word frequencies.

# Step 4: Logic
# Print the first word with frequency one and stop.

# Step 5: Code
sentence = "apple banana apple mango banana orange"
words = sentence.lower().split()
for word in words:
    if words.count(word) == 1:
        print(word)
        break

# Step 6: Test
test_sentence = "cat dog cat bird"
test_words = test_sentence.lower().split()
for word in test_words:
    if test_words.count(word) == 1:
        print(word)
        break
```
